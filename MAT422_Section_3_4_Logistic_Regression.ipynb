```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

np.set_printoptions(precision=4, suppress=True)
```

## Sigmoid/logistic function

This section gives one compact Python demonstration of **Sigmoid/logistic function**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z = np.linspace(-15, 15, 300)
plt.plot(z, sigmoid(z))
plt.axhline(0.5, color="black", linewidth=0.8)
plt.xlabel("z")
plt.ylabel("sigmoid(z)")
plt.title("Logistic function maps real numbers to probabilities")
plt.show()
```

The sigmoid curve stays S-shaped no matter the z-range and it’s always squeezed between 0 and 1 and crosses 0.5 exactly at z=0. Widening the range just shows the flat tails more clearly, where the function saturates near 0 or 1 for large |z|.

## Binary classification probabilities

This section gives one compact Python demonstration of **Binary classification probabilities**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
X, y = make_classification(n_samples=200, n_features=2, n_redundant=0, n_informative=2,
                            n_clusters_per_class=1, class_sep=0.6, random_state=422)
model = LogisticRegression()
model.fit(X, y)
pred = model.predict(X)
proba = model.predict_proba(X)[:, 1]

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("Training accuracy:", accuracy_score(y, pred))
print("Confusion matrix:")
print(confusion_matrix(y, pred))
```

With the classes closer together, the model’s training accuracy dropped compared to a more separate version, and the confusion matrix shows more errors. That makes sense since logistic regression draws a single linear boundary, so when the classes overlap it can’t perfectly separate them anymore.

## Logistic regression fitting and interpretation

This section gives one compact Python demonstration of **Logistic regression fitting and interpretation**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
xx, yy = np.meshgrid(np.linspace(X[:,0].min()-1, X[:,0].max()+1, 300),
                      np.linspace(X[:,1].min()-1, X[:,1].max()+1, 300))
grid = np.c_[xx.ravel(), yy.ravel()]
zz = model.predict_proba(grid)[:, 1].reshape(xx.shape)

plt.contourf(xx, yy, zz, levels=20, cmap="PiYG", alpha=0.6)
plt.colorbar(label="P(class 1)")
plt.scatter(X[:,0], X[:,1], c=y, edgecolor="black", cmap="bwr")
plt.title("Logistic regression probability surface")
plt.show()
```

The contour plot shows the predicted probability of class 1 across the input space, with the color changing smoothly from one class’s region to the other. With the classes closer together in the previous step, the probability surface has a wider band near the boundary instead of a sharp transition, which shows that there is more uncertainty in that region.
