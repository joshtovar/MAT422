```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
rng = np.random.default_rng(422)
```

## Joint probability distributions

This section gives one compact Python demonstration of **Joint probability distributions**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
#X and Y are independent here
joint = np.array([[0.06, 0.18, 0.16],
                  [0.09, 0.27, 0.24]])
x_values = np.array([0, 1])
y_values = np.array([0, 1, 2])

px = joint.sum(axis=1)
py = joint.sum(axis=0)
print("Joint total =", joint.sum())
print("P(X) =", px)
print("P(Y) =", py)
print("P(Y=2 | X=1) =", joint[1, 2] / px[1])
```

P(Y=2 | X=1) equals P(Y=2) here, which means knowing X doesn't tell us anything about Y. That's what independence is. The conditional probability is the same as the marginal, and every cell in the joint table is just the product of its row and column marginals.

## Correlation and dependence

This section gives one compact Python demonstration of **Correlation and dependence**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
#Flips to negative slope
x = rng.normal(0, 1, 200)
y_dependent = -3 * x + rng.normal(0, 1.5, 200)
y_independent = rng.normal(0, 1, 200)

print("corr(x, dependent y) =", np.corrcoef(x, y_dependent)[0, 1])
print("corr(x, independent y) =", np.corrcoef(x, y_independent)[0, 1])

plt.scatter(x, y_dependent, alpha=0.7, label="dependent")
plt.scatter(x, y_independent, alpha=0.5, label="mostly independent")
plt.legend()
plt.title("Correlation as a dependence summary")
plt.show()
```

The correlation for y_dependent came out around -0.89. The negative sign means y goes down as x goes up, and the size (close to 1) means the points sit pretty tight around a line. I added more noise than before, so it's not as strong as it would be with less noise. y_independent is near 0 because it has no linear link to x. Correlation measures the direction and strength of a linear relationship.

## Random samples

This section gives one compact Python demonstration of **Random samples**. After running the code, add your own interpretation and change at least one input, parameter, or example so the notebook reflects your own work.


```python
#Increase sample size
sample_means = []
for _ in range(1000):
    sample = rng.exponential(scale=2.0, size=100)
    sample_means.append(sample.mean())

sample_means = np.array(sample_means)
print("Mean of sample means =", sample_means.mean())
print("Std of sample means =", sample_means.std())

plt.hist(sample_means, bins=30, color="tab:green", edgecolor="white")
plt.title("Sampling distribution of the sample mean")
plt.xlabel("sample mean")
plt.ylabel("count")
plt.show()
```

Each sample mean is the average of n draws, and when you repeat that 1000 times you get the sampling distribution of the mean. Its center stays at the true mean (2), but the spread shrinks like σ/√n, so going from n=30 to n=100 took the std from about 0.37 down to about 0.2
